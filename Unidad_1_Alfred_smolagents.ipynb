```python
from google.colab import userdata
from smolagents import CodeAgent, OpenAIModel, tool
from datetime import datetime
from zoneinfo import ZoneInfo

model = OpenAIModel(
    model_id="gemini-3.5-flash-lite",
    api_base="https://generativelanguage.googleapis.com/v1beta/openai/",
    api_key=userdata.get("GEMINI_API_KEY")
)
```

```python
@tool
def get_current_time_in_timezone(timezone: str) -> str:
    """
    Returns the current time for an IANA timezone.

    Args:
        timezone: An IANA timezone such as America/Bogota.
    """
    try:
        current = datetime.now(ZoneInfo(timezone))
        return current.strftime("%Y-%m-%d %H:%M:%S %Z")
    except Exception as e:
        return f"Invalid timezone: {e}"
```

```python
alfred = CodeAgent(
    tools=[get_current_time_in_timezone],
    model=model,
    max_steps=4
)

resultado = alfred.run(
    "What is the current time in Bogota, Colombia? "
    "Use the timezone tool."
)

print("Alfred:", resultado)
```
