```python
from google.colab import userdata
from smolagents import OpenAIModel

api_key = userdata.get("GEMINI_API_KEY")

model = OpenAIModel(
    model_id="gemini-3.5-flash-lite",
    api_base="https://generativelanguage.googleapis.com/v1beta/openai/",
    api_key=api_key
)

print("Modelo configurado correctamente")
```

```python
respuesta = model.generate([
    {
        "role": "user",
        "content": "Explain in one sentence what an AI agent is."
    }
])

print(respuesta.content)
```

```python
from smolagents import CodeAgent, tool

@tool
def calcular_promedio(numeros: list[float]) -> float:
    """
    Calcula el promedio de una lista de números.

    Args:
        numeros: Lista de números para calcular el promedio.
    """
    return sum(numeros) / len(numeros)

agente = CodeAgent(
    tools=[calcular_promedio],
    model=model,
    max_steps=3
)

resultado = agente.run(
    "Calcula el promedio de 8, 10, 15 y 19 usando la herramienta calcular_promedio."
)

print("Resultado final:", resultado)
```
